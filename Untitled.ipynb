```python
import pandas as pd

# Load without low_memory=False first — only add that if you hit a warning
works = pd.read_csv("Works Sanctioned.csv")

works.columns = works.columns.str.strip()
print(works.columns.tolist())
print("Total rows before cleaning:", len(works))
```

```python
# Convert dates
works['Recommended date'] = pd.to_datetime(works['Recommended date'], format='%d-%b-%Y', errors='coerce')
works['Sanction Date'] = pd.to_datetime(works['Sanction Date'], format='%d-%b-%Y', errors='coerce')

# Convert Sanction Amount to numeric
works['Sanction Amount ( ₹ )'] = pd.to_numeric(works['Sanction Amount ( ₹ )'], errors='coerce')

# Rename for easier typing
works = works.rename(columns={'Sanction Amount ( ₹ )': 'Sanction_Amount'})

# Drop rows missing critical fields (removes Grand Total row and any other bad rows)
works_clean = works.dropna(subset=['Recommended date', 'Sanction Date', 'Sanction_Amount'])

print("Rows remaining after cleaning:", len(works_clean))
print(works_clean.dtypes)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# Create the flag column
works_clean['is_duplicate_flag'] = False

print("Running TF-IDF duplicate detection by MP (this might take ~30-60 seconds)...")

# Group by MP to limit matrix size and find localized fraud
for mp, group in works_clean.groupby("Hon'ble Members of Parliament"):
    if len(group) < 2:
        continue # Can't have duplicates with only 1 project

    # 1. Vectorize text (convert words to numbers based on importance)
    vectorizer = TfidfVectorizer(stop_words='english')
    try:
        # Fill missing text with empty string to prevent crashes
        tfidf_matrix = vectorizer.fit_transform(group['Work description'].fillna(''))
        
        # 2. Compute similarity matrix
        sim_matrix = cosine_similarity(tfidf_matrix)

        # 3. Ignore exact self-matches (the diagonal of the matrix)
        np.fill_diagonal(sim_matrix, 0)
        
        # 4. Find rows with > 85% similarity to ANY other row in this MP's portfolio
        has_duplicate = (sim_matrix > 0.85).any(axis=1)

        # Map the flagged rows back to the main dataframe
        duplicate_indices = group.index[has_duplicate]
        works_clean.loc[duplicate_indices, 'is_duplicate_flag'] = True
    except ValueError:
        # Catches edge cases (like a group of completely blank descriptions)
        continue

print(f"Total potential duplicate descriptions flagged: {works_clean['is_duplicate_flag'].sum()}")
```

```python
from sklearn.ensemble import IsolationForest

print("\n--- Running Isolation Forest (ML Anomaly Detection) ---")

# 1. Select the numeric features
# We use absolute value for z-score (extremely cheap and extremely expensive are both weird)
ml_features = works_clean[['days_to_sanction', 'cost_zscore', 'days_since_sanction']].copy()
ml_features['cost_zscore'] = ml_features['cost_zscore'].abs()

# 2. Impute Missing Values (ML models crash if they see 'NaN')
# Fill missing z-scores with 0 (meaning 'exactly average')
ml_features['cost_zscore'] = ml_features['cost_zscore'].fillna(0)
# Fill missing dates with the median
ml_features['days_to_sanction'] = ml_features['days_to_sanction'].fillna(ml_features['days_to_sanction'].median())
ml_features['days_since_sanction'] = ml_features['days_since_sanction'].fillna(ml_features['days_since_sanction'].median())

# 3. Initialize the model
# contamination=0.05 tells the model: "Find the 5% weirdest rows in the whole dataset"
iso_forest = IsolationForest(contamination=0.05, random_state=42)

# 4. Fit and predict (-1 means anomaly, 1 means normal)
works_clean['iforest_prediction'] = iso_forest.fit_predict(ml_features)

# Convert to a clean True/False flag
works_clean['is_ml_anomaly'] = works_clean['iforest_prediction'] == -1

ml_anomalies_count = works_clean['is_ml_anomaly'].sum()
print(f"Isolation Forest flagged {ml_anomalies_count} records as multi-dimensional anomalies.")

# 5. Find the "Sneaky" Anomalies
# These are rows the ML caught, but our manual risk score (from earlier) missed
sneaky_anomalies = works_clean[(works_clean['is_ml_anomaly'] == True) & (works_clean['Risk_Score'] < 2)]
print(f"Found {len(sneaky_anomalies)} 'sneaky' anomalies that bypassed the strict manual rules.")
```

```python
from sklearn.ensemble import IsolationForest

print("--- Running Isolation Forest (ML Anomaly Detection) ---")

# 1. Select the numeric features and clean them for the ML model
ml_features = works_clean[['days_to_sanction', 'cost_zscore', 'days_since_sanction']].copy()
ml_features['cost_zscore'] = ml_features['cost_zscore'].abs().fillna(0)
ml_features['days_to_sanction'] = ml_features['days_to_sanction'].fillna(ml_features['days_to_sanction'].median())
ml_features['days_since_sanction'] = ml_features['days_since_sanction'].fillna(ml_features['days_since_sanction'].median())

# 2. Run the Model (Looking for the top 5% weirdest combinations)
iso_forest = IsolationForest(contamination=0.05, random_state=42)
works_clean['is_ml_anomaly'] = iso_forest.fit_predict(ml_features) == -1

print(f"Isolation Forest flagged {works_clean['is_ml_anomaly'].sum()} records as complex anomalies.")

# 3. Create the Final Master 'Risk Score' (Out of 4 possible red flags)
works_clean['Total_Risk_Score'] = (
    works_clean['is_cost_outlier'].astype(int) + 
    works_clean['is_speed_outlier'].astype(int) + 
    works_clean['is_stalled'].astype(int) + 
    works_clean['is_ml_anomaly'].astype(int)
)

print("\n--- Final Project Risk Distribution ---")
print(works_clean['Total_Risk_Score'].value_counts().sort_index(ascending=False))

# 4. EXPORT THE DATASET
# We save this so your Streamlit frontend can just load this clean file
export_filename = 'sih_audited_works_final.csv'
works_clean.to_csv(export_filename, index=False)
print(f"\nSUCCESS! Cleaned and scored dataset saved to: {export_filename}")
```
